```python
# 24BAD006
# Ajay Krishnan T

import numpy as np
import matplotlib.pyplot as plt

# Generate input values
x = np.linspace(-10, 10, 500)

# Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Calculate outputs
y_sigmoid = sigmoid(x)
y_tanh = tanh(x)
y_relu = relu(x)

# Plot Activation Functions
plt.figure(figsize=(10,6))

plt.plot(x, y_sigmoid, label='Sigmoid', linewidth=2)
plt.plot(x, y_tanh, label='Tanh', linewidth=2)
plt.plot(x, y_relu, label='ReLU', linewidth=2)

plt.title("Comparison of Activation Functions")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Performance Comparison of Activation Functions

# Ajay Krishnan T
# 24BAD006


import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
import matplotlib.pyplot as plt
import pandas as pd

# Load MNIST Dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize data
x_train = x_train / 255.0
x_test = x_test / 255.0

# Function to create ANN model
def create_model(activation):
    model = Sequential([
        Flatten(input_shape=(28,28)),
        Dense(128, activation=activation),
        Dense(64, activation=activation),
        Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model

# Activation functions to compare
activations = ['sigmoid', 'tanh', 'relu']

results = {}

for act in activations:

    print("\n======================================")
    print("Training Model with:", act.upper())
    print("======================================")

    model = create_model(act)

    history = model.fit(
        x_train,
        y_train,
        epochs=10,
        batch_size=128,
        validation_data=(x_test, y_test),
        verbose=1
    )

    results[act] = history.history

# Display Comparison Table
comparison = []

for act in activations:
    history = results[act]

    comparison.append({
        "Activation Function": act.upper(),
        "Training Accuracy": round(history['accuracy'][-1],4),
        "Validation Accuracy": round(history['val_accuracy'][-1],4),
        "Training Loss": round(history['loss'][-1],4),
        "Validation Loss": round(history['val_loss'][-1],4),
        "Epochs": len(history['accuracy'])
    })

df = pd.DataFrame(comparison)

print("\n")
print("="*70)
print("FINAL PERFORMANCE COMPARISON")
print("="*70)

print(df)

# Plot Validation Accuracy
plt.figure(figsize=(10,5))

for act in activations:
    plt.plot(results[act]['val_accuracy'], label=act.upper())

plt.title("Validation Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Plot Validation Loss
plt.figure(figsize=(10,5))

for act in activations:
    plt.plot(results[act]['val_loss'], label=act.upper())

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

```

```python
#24BAD006
#Ajay Krishnan T

import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import SGD, RMSprop, Adam
import matplotlib.pyplot as plt
import pandas as pd

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0

def model():
    m = Sequential([
        Flatten(input_shape=(28,28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])
    return m

optimizers = {
    "SGD": SGD(),
    "Momentum": SGD(momentum=0.9),
    "RMSProp": RMSprop(),
    "Adam": Adam()
}

history_data = {}
results = []

for name, opt in optimizers.items():
    m = model()
    m.compile(optimizer=opt,
              loss="sparse_categorical_crossentropy",
              metrics=["accuracy"])

    history = m.fit(
        x_train, y_train,
        epochs=10,
        batch_size=128,
        validation_data=(x_test, y_test),
        verbose=1
    )

    history_data[name] = history.history

    results.append([
        name,
        history.history["loss"][-1],
        history.history["val_loss"][-1],
        history.history["accuracy"][-1],
        history.history["val_accuracy"][-1],
        len(history.history["loss"])
    ])

df = pd.DataFrame(results, columns=[
    "Optimizer",
    "Training Loss",
    "Validation Loss",
    "Training Accuracy",
    "Validation Accuracy",
    "Epochs"
])

print(df)

plt.figure(figsize=(10,5))
for name in optimizers:
    plt.plot(history_data[name]["loss"], label=name)
plt.title("Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(10,5))
for name in optimizers:
    plt.plot(history_data[name]["accuracy"], label=name)
plt.title("Accuracy vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()
```
